```python
import requests
from bs4 import BeautifulSoup
from fake_useragent import UserAgent
```

```python
ua = UserAgent()
target_url = "http://books.toscrape.com"
fake_header = {

    "User_agent":ua.random
}
response = requests.get(target_url,headers=fake_header)
response.encoding = response.apparent_encoding
```

```python
print(response.status_code)
```

```python
raw_html_text = response.text
print("Successfully data loaded")
```

```python
soup = BeautifulSoup(raw_html_text,"html.parser")
all_books = soup.find_all("article")
```

```python
print(len(all_books))
```

```python
for index,book in enumerate(all_books,1):
    book_h3 = book.find("h3") 
    book_link = book_h3.find("a")
    book_price = book.find(class_= "price_color")
    print(index,"-","Book title:",book_link['title'] ,"| \nPrice:" , book_price.text)
   
```

```python
import csv
with open("book_prices.csv","w" ,newline="",encoding='utf-8-sig') as file:
    writer = csv.writer(file)
    
    writer.writerow(['Title ID','Book Title','Product price'])
    for index, book in enumerate(all_books, 1):
        book_h3 = book.find("h3") 
        book_link = book_h3.find("a")
        book_price = book.find(class_="price_color")
        writer.writerow([index, book_link['title'], book_price.text])
print('Spreadsheet Successfully created!')
```
